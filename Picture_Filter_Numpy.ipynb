### Dominant Color Masks

This visualization shows the original image alongside three binary masks.  
Each mask highlights pixels where one color channel is dominant over the other two:  
- **Red dominant**: pixels where Red > Green and Red > Blue  
- **Green dominant**: pixels where Green > Red and Green > Blue  
- **Blue dominant**: pixels where Blue > Red and Blue > Green  

The plots also display the count of dominant pixels for each color, making it easy to compare color distribution across the image.


```python
y = x.copy()

mask_red = (y[:, :, 0] > y[:, :, 1]) & (y[:, :, 0] > y[:, :, 2])
mask_green = (y[:, :, 1] > y[:, :, 2]) & (y[:, :, 1] > y[:, :, 0])
mask_blue = (y[:, :, 2] > y[:, :, 1]) & (y[:, :, 2] > y[:, :, 0])

px_mask_red = np.sum(mask_red)  # the count of mask_red
px_mask_green = np.sum(mask_green)  # the count of mask_green
px_mask_blue = np.sum(mask_blue)  # the count of mask_blue

fig, axes = plt.subplots(2, 2, figsize=(10, 10))

axes[0,0].imshow(x)
axes[0,0].set_title("Original", fontsize=10)
axes[0,0].axis("off")

axes[0,1].imshow(mask_red)
axes[0,1].set_title("(R > G & B) px:"+str(int(px_mask_red)), fontsize=10)
axes[0,1].axis("off")

axes[1,0].imshow(mask_green)
axes[1,0].set_title("(G > R & B) px:"+str(int(px_mask_green)), fontsize=10)
axes[1,0].axis("off")

axes[1,1].imshow(mask_blue)
axes[1,1].set_title("(B > R & G) px:"+str(int(px_mask_blue)), fontsize=10)
axes[1,1].axis("off")

plt.tight_layout()
plt.show()

```

#### In this section, we observe that the number of selected pixels is less than the total number of pixels in the image. This issue arises because of the conditions we defined for pixel selection. For example, pixels where the values of R, G, and B are equal were not included. To ensure that all pixels are selected, the boundary cases where R, G, and B values are equal must also be incorporated into the conditions.

### Dominant Color Filter

This code applies a filter where, for each pixel, only the dominant RGB channel 
(highest value among Red, Green, and Blue) is preserved. The other channels are 
set to zero, resulting in an image that highlights the strongest color at each pixel.


```python

y = x.copy()

max_channel = np.argmax(y, axis=2)

filtered = np.zeros_like(y)

rows, cols = y.shape[0], y.shape[1]
for c in range(3):
    mask = (max_channel == c)
    filtered[mask, c] = y[mask, c]

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(x)
axes[0].set_title("Original", fontsize=10)
axes[0].axis("off")

axes[1].imshow(filtered)
axes[1].set_title("Dominant color per pixel", fontsize=10)
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

### Negative Image Filter

This code inverts the colors of the image by subtracting each RGB value from 255.  
For example, a pixel with (R=20, G=100, B=200) becomes (R=235, G=155, B=55).  
The result is a negative version of the original image, highlighting complementary colors.


```python

y = x.copy()

y = 255 - y

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(x)
axes[0].set_title("Original", fontsize=10)
axes[0].axis("off")

axes[1].imshow(y)
axes[1].set_title("Negative Image", fontsize=10)
axes[1].axis("off")

plt.tight_layout()
plt.show()

```

### Original vs Negative with Row-wise Color Averages

This visualization shows the original image and its negative version side by side.  
Below each image, line plots display the row-wise average intensities of Red, Green, and Blue channels.  
The charts highlight how color distribution changes from top to bottom in both the original and the negative image.


```python
y = 255 - x

def row_means(img):
    red   = img[:, :, 0].mean(axis=1)
    green = img[:, :, 1].mean(axis=1)
    blue  = img[:, :, 2].mean(axis=1)
    return red, green, blue

red_orig, green_orig, blue_orig = row_means(x)
red_neg, green_neg, blue_neg    = row_means(y)

rows = np.arange(x.shape[0])

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0,0].imshow(x)
axes[0,0].set_title("Original", fontsize=12)
axes[0,0].axis("off")

axes[1,0].plot(rows, red_orig, color='red', label='Red')
axes[1,0].plot(rows, green_orig, color='green', label='Green')
axes[1,0].plot(rows, blue_orig, color='blue', label='Blue')
axes[1,0].set_title("Row-wise averages (Original)", fontsize=12)
axes[1,0].set_xlabel("Row index")
axes[1,0].set_ylabel("Average intensity")
axes[1,0].legend()
axes[1,0].grid(True)

axes[0,1].imshow(y)
axes[0,1].set_title("Negative", fontsize=12)
axes[0,1].axis("off")

axes[1,1].plot(rows, red_neg, color='red', label='Red')
axes[1,1].plot(rows, green_neg, color='green', label='Green')
axes[1,1].plot(rows, blue_neg, color='blue', label='Blue')
axes[1,1].set_title("Row-wise averages (Negative)", fontsize=12)
axes[1,1].set_xlabel("Row index")
axes[1,1].set_ylabel("Average intensity")
axes[1,1].legend()
axes[1,1].grid(True)

plt.tight_layout()
plt.show()
```

### Width Expansion with `repeat` *2

This code expands the image horizontally by repeating each column of pixels.  
The `repeat` function creates a step-like duplication, so the image width grows while columns are duplicated.  
As a result, the output shows repeated columns, giving the impression of a stretched image with stair-like patterns.


```python
y = x.copy()

y_upscale_w = np.repeat(y,2,axis=1)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(x)
axes[0].set_title("Original", fontsize=10)
axes[0].axis("off")

axes[1].imshow(y_upscale_w)
axes[1].set_title("Width upscale", fontsize=10)
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
y_upscale_w.shape
```

### Height Expansion with `repeat` *2

This code expands the image vertically by repeating each row of pixels.  
The `repeat` function creates step-like duplication, so the image height grows while rows are duplicated.  
As a result, the output shows repeated rows, giving the impression of a stretched image with stair-like patterns.


```python
y = x.copy()

y_upscale_h = np.repeat(y,2,axis=0)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(x)
axes[0].set_title("Original", fontsize=10)
axes[0].axis("off")

axes[1].imshow(y_upscale_h)
axes[1].set_title("Height upscale", fontsize=10)
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
y_upscale_h.shape
```

### Width and Height Expansion with `repeat`

This code expands the image in both directions by repeating rows and columns of pixels.  
The `repeat` function duplicates data along both axes, so the image grows larger while creating stair-like patterns.  
As a result, the output shows repeated rows and columns, producing a stretched version of the original image.


```python
y = x.copy()

y_upscale_w_h = np.repeat(np.repeat(y,2,axis=0,),2,axis=1)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(x)
axes[0].set_title("Original", fontsize=10)
axes[0].axis("off")

axes[1].imshow(y_upscale_w_h)
axes[1].set_title("Height & Width upscale", fontsize=10)
axes[1].axis("off")

plt.tight_layout()
plt.show()
```
